### Problem 1
Generate a 2-dimensional Numpy array of shape (15,15) filled with random integers between 0 and 255(inclusive). Perform the following tasks:
1. Create a new array where all the elements less than the threshold(120) are set to zero and the rest of them remain the same.
2. Create a Python function named `boundarytransversal(matrix)` that will print out boundary elements of the matrix starting from the top-left corner and in clockwise direction.
3. Create a new Numpy array by replacing all the odd elements with zero and even with one.

```python
# problem 1
matrix_1 = np.random.randint(0, 256, size=(15, 15))
print(matrix_1)

# Task 1.1
matrix_thresh = np.where(matrix_1 < 120, 0, matrix_1)
print(matrix_thresh)
```

```python
# Task 1.2
def boundary_transversal(matrix):
    if len(matrix) == 0:
        return
    
    rows, cols = matrix.shape
    boundary = []
    
    # Top row (left to right)
    for j in range(cols):
        boundary.append(matrix[0, j])
        
    # Right column (top to bottom)
    for i in range(1, rows):
        boundary.append(matrix[i, cols - 1])
        
    # Bottom row (right to left)
    if rows > 1:
        for j in range(cols - 2, -1, -1):
            boundary.append(matrix[rows - 1, j])
            
    # Left column (bottom to top)
    if cols > 1:
        for i in range(rows - 2, 0, -1):
            boundary.append(matrix[i, 0])
            
    print("Boundary Elements :", boundary)

boundary_transversal(matrix_1)
boundary_transversal(matrix_thresh)
```

```python
# Task 1.3: odd-> 0, even -> 1
matrix_even_odd = np.where(matrix_1 % 2 != 0, 0, 1)
print(matrix_even_odd)
```

### Problem 2
Generate a 2-dimensional numpy array of size (20,20) with random integers between 0 and 255(inclusive), then perform the following tasks:
1. Calculate and print the maximum, minimum, and median of each row and column separately.
2. Generate and display the image that has to be created using the above Numpy array after resizing it to (200,200). Use the nearest neighbor interpolation that enlarges each pixel to a (10x10) square of the same value, and no smoothing or averaging is applied to that.
3. Create 2 grayscale images, one with a horizontal gradient (0 to 255 left to right) and another with a vertical gradient( 0 to 255 top to bottom) of size (256,256) pixels, and display them.

```python
# random 2-d numpy array
matrix_2 = np.random.randint(0, 256, size=(20, 20)).astype(np.uint8)
print(matrix_2)
```

```python

# Task 2.1: Max, Min, Median of each row and column
row_max = np.max(matrix_2, axis=1)
row_min = np.min(matrix_2, axis=1)
row_median = np.median(matrix_2, axis=1)

col_max = np.max(matrix_2, axis=0)
col_min = np.min(matrix_2, axis=0)
col_median = np.median(matrix_2, axis=0)

print("Row Max:", row_max)
print("Row Min:", row_min)
print("Row Median:", row_median)
print("Col Max:", col_max)
print("Col Min:", col_min)
print("Col Median:", col_median)
```

```python
import cv2

img_resized = cv2.resize(matrix_2, (200, 200), interpolation=cv2.INTER_NEAREST)

plt.figure(figsize=(10, 4))
plt.subplot(1, 3, 1)
plt.title("Resized")
plt.imshow(img_resized, cmap='gray')
```

```python
# Task 2.3: Horizontal and vertical gradient grayscale images of size (256, 256)
horizontal_grad = np.tile(np.arange(256, dtype=np.uint8), (256, 1))
vertical_grad = np.tile(np.arange(256, dtype=np.uint8).reshape(256, 1), (1, 256))

plt.subplot(1, 3, 2)
plt.title("Horizontal Gradient")
plt.imshow(horizontal_grad, cmap='gray')
plt.axis('off')

plt.subplot(1, 3, 3)
plt.title("Vertical Gradient")
plt.imshow(vertical_grad, cmap='gray')
plt.axis('off')

plt.tight_layout()
plt.show()
```

### Problem 3
Read an image in Grayscale format and perform the following operations:
1. Invert the image (invert all pixel values) and display it.
Hint:For an image with uint8 datatype, inversion = 255 - image.
2. Read an image in colored format and:
 - Split it into R, G, B channels
 - Zero out the Red channel
 - Merge the channels and display the modified image
3. Read an image in color format and:
 - flip it horizontally and display it.
 - flip it vertically and display it.
 - Mirror the image and display it. (Think of the effect of mirroring an image, considering an image as a 2D matrix, is it the same as horizontal flipping ?)

```python
# 3.1 Create a dummy grayscale image
img_gray = np.random.randint(0, 256, (256, 256), dtype=np.uint8)

# Invert the grayscale image
img_inv = 255 - img_gray


plt.figure(figsize=(8, 4))

plt.subplot(1, 2, 1)
plt.title("Original Grayscale")
plt.imshow(img_gray, cmap='gray')

plt.subplot(1, 2, 2)
plt.title("Inverted Grayscale")
plt.imshow(img_inv, cmap='gray')

plt.tight_layout()
plt.show()
```

```python
# 3.2 Create a dummy color image (256x256x3)
img_color = np.random.randint(0, 256, (256, 256, 3), dtype=np.uint8)

# Split BGR channels 
b, g, r = cv2.split(img_color)
# Zero out the red channel
r_zero = np.zeros_like(r)
# Merge channel
img_no_red = cv2.merge((b, g, r_zero))

# Convert BGR to RGB for Matplotlib
img_color_rgb = cv2.cvtColor(img_color, cv2.COLOR_BGR2RGB)
img_no_red_rgb = cv2.cvtColor(img_no_red, cv2.COLOR_BGR2RGB)

# Display
plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.title("Original Color")
plt.imshow(img_color_rgb)

plt.subplot(1, 2, 2)
plt.title("Red Channel Removed")
plt.imshow(img_no_red_rgb)

plt.tight_layout()
plt.show()
```

```python
# 3.3 Flip horizontally, vertically, and mirror
if img_color is not None:
    img_rgb = cv2.cvtColor(img_color, cv2.COLOR_BGR2RGB)
    
    # Flip horizontally 
    img_flip_h = cv2.flip(img_rgb, 1)
    # Flip vertically
    img_flip_v = cv2.flip(img_rgb, 0)
    
    plt.figure(figsize=(12, 4))
    plt.subplot(1, 3, 1)
    plt.title("Original")
    plt.imshow(img_rgb)
    
    plt.subplot(1, 3, 2)
    plt.title("Horizontal Flip")
    plt.imshow(img_flip_h)

    
    plt.subplot(1, 3, 3)
    plt.title("Vertical Flip")
    plt.imshow(img_flip_v)

    plt.show()
```

When an image is treated as a 2D matrix, mirroring across the vertical axis is equivalent to horizontal flipping

### Problem 4
Read a colored image and do the following operations:
1. Display the size of the image and the pixel values of different channels. Note that cv2.imshow() read image in BGR format.
2. Write a Python function that will manually convert a Color image to a Grayscale image. Don't use cv2.cvtColor() function. Note that, Gray = 0.299R + 0.587G + 0.114B
3. Count the number of Pure white pixels( have value [255, 255, 255] in a color image.

```python

if img_color is not None:
    # 4.1 Display size and pixel values 
    print(f"Image Size (Height, Width, Channels): {img_color.shape}")
    b, g, r = cv2.split(img_color)
    print("Blue Channel values:")
    print(b[:3, :3])
    print("Green Channel values:")
    print(g[:3, :3])
    print("Red Channel values:")
    print(r[:3, :3])
    
    # 4.2 Manual color to grayscale 
    def manual_rgb_to_gray(image_bgr):
        B = image_bgr[:, :, 0]
        G = image_bgr[:, :, 1]
        R = image_bgr[:, :, 2]
        
        # Gray = 0.299R + 0.587G + 0.114B
        gray = 0.299 * R + 0.587 * G + 0.114 * B
        return gray.astype(np.uint8)

    gray_manual = manual_rgb_to_gray(img_color)
    
    plt.figure(figsize=(4, 4))
    plt.title("Manual Grayscale")
    plt.imshow(gray_manual, cmap='gray')
    plt.show()
    
    # 4.3 Count number of pure white pixels
    img_color[10:20, 10:20] = [255, 255, 255]
    
    # In BGR, white is [255, 255, 255]
    white_pixels_mask = (img_color[:, :, 0] == 255) & (img_color[:, :, 1] == 255) & (img_color[:, :, 2] == 255)
    pure_white_count = np.sum(white_pixels_mask)
    print(f"Number of pure white pixels: {pure_white_count}")
```

### Problem 5
Read a color image and perform the following operations:
1. Convert it to a grayscale image using OpenCV, and apply a Gaussian Blur with a kernel size of (9,9) and display it.
2. Create a white image of the same size and subtract the blurred grayscale image from the white image. Display the resulting image. Note that this effect is same as the inversion of an image.
3. create a blank image of size (512,512) and do the following:
 - Create a rectangle of blue color with top-left corner at (50,50) and bottom-right corner at (462,462), with a thickness of 3 pixels.
 - Draw a circle of radius = 50, centre at (300,300) and filled with green color.
 - Draw a red diagonal line starting from the top-right to the bottom left corner of thickness = 2 pixels.

```python

if img_color is not None:
    # 5.1 Converting to grayscale and  Gaussian Blur
    gray_cv = cv2.cvtColor(img_color, cv2.COLOR_BGR2GRAY)
    blurred = cv2.GaussianBlur(gray_cv, (9, 9), 0)
    
    plt.figure(figsize=(8, 4))
    plt.subplot(1, 2, 1)
    plt.title("Blurred Grayscale")
    plt.imshow(blurred, cmap='gray')
    
    # 5.2 Subtract blurred from white image
    white_img = np.ones_like(blurred) * 255
    subtracted = cv2.subtract(white_img, blurred)
    
    plt.subplot(1, 2, 2)
    plt.title("White-Blurred (Inverted)")
    plt.imshow(subtracted, cmap='gray')
    plt.show()
```

```python
# 5.3 Create blank image and draw shapes
# Blank image 
blank_img = np.zeros((512, 512, 3), dtype=np.uint8)

# Rectangle of blue color
cv2.rectangle(blank_img, (50, 50), (462, 462), (255, 0, 0), 3)

# Circle filled with green color
cv2.circle(blank_img, (300, 300), 50, (0, 255, 0), -1)

# Red diagonal line
cv2.line(blank_img, (511, 0), (0, 511), (0, 0, 255), 2)

blank_rgb = cv2.cvtColor(blank_img, cv2.COLOR_BGR2RGB)
plt.figure(figsize=(6, 6))
plt.title("Shapes on Blank Image")
plt.imshow(blank_rgb)
plt.show()
```
