### Problem 1: 
Use the YOLOv8 nano model to perform real-time object detection using your webcam.
Tasks:
1. Load the yolov8n.pt model.
2. Capture a live feed from your webcam using OpenCV.
3. Run real-time detection frame-by-frame and show the results live with bounding boxes.
4. Press q to quit the window.

````python
from ultralytics import YOLO
import cv2

# 1. Load the yolov8n.pt model
model = YOLO("yolov8n.pt")
````

````python
# 2. Capture live feed from webcam
cap = cv2.VideoCapture(0)

if not cap.isOpened():
    print("Error: Could not open webcam.")
else:
    while True:
        ret, frame = cap.read()
        if not ret:
            print("Failed to grab frame")
            break
            
        # 3. Run real-time detection frame-by-frame
        results = model.predict(frame, stream=True)
        
        # Show results live with bounding boxes
        for result in results:
            annotated_frame = result.plot()
            
        cv2.imshow("YOLOv8 Real-time Detection", annotated_frame)
        
        # 4. Press 'q' to quit the window
        if cv2.waitKey(1) & 0xFF == ord('q'):
            break

    cap.release()
    cv2.destroyAllWindows()
````

**Training time**:
The model was trained for 10 epochs and completed in approximately 8.7 seconds.

### Problem 5: 
Use the Ultralytics YOLOv8 prediction results to extract and visualize detailed information.
Tasks:
1. For a single prediction image, extract the following:
- Bounding box coordinates
- Class label
- Confidence score
2. Print the prediction information.
3. Draw custom bounding boxes and labels on the image using OpenCV (without `results.show()`).

📌 Hint: Use `results[0].boxes.xyxy`, `results[0].boxes.conf`, and `results[0].boxes.cls`.

````python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from ultralytics import YOLO


img_test = "custom_data/images/test/test_2.jpeg"

# Load model and predict
model = YOLO("yolov8n.pt")
results = model.predict(img_test)

# Load original image for drawing
img_to_draw = cv2.imread(img_test)
result = results[0] 



# 1. Extract and 2. Print
if len(result.boxes) == 0:
    print("No objects detected in the image. (Expected behavior for random noise)")
else:
    for box in result.boxes:
        # Extract coordinates, confidence, and class id
        xyxy = box.xyxy[0].cpu().numpy() # [x1, y1, x2, y2]
        conf = box.conf[0].cpu().item()
        cls_id = int(box.cls[0].cpu().item())
        class_name = model.names[cls_id]
        
        print(f"Detected: {class_name}")
        print(f"Bounding Box (x1, y1, x2, y2): {xyxy}")
        print(f"Confidence: {conf:.2f}\n")
        
        # 3. Draw custom bounding boxes and labels using OpenCV
        x1, y1, x2, y2 = map(int, xyxy)
        
        # Draw Rectangle
        cv2.rectangle(img_to_draw, (x1, y1), (x2, y2), (0, 0, 255), 2)
        
        # Draw Label
        label_text = f"{class_name} {conf:.2f}"
        cv2.putText(img_to_draw, label_text, (x1, max(y1 - 10, 10)), 
                    cv2.FONT_HERSHEY_SIMPLEX, 0.5, (0, 0, 255), 2)

# Display final custom annotated image
plt.figure(figsize=(6, 6))
plt.title("Custom Drawn Bounding Boxes")
plt.imshow(cv2.cvtColor(img_to_draw, cv2.COLOR_BGR2RGB))
plt.axis('off')
plt.show()
````
